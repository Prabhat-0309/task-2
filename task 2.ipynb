```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'].dropna(), bins=50, kde=True, color='skyblue')
plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Fare'], bins=60, kde=True, color='skyblue')
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.show()
```

```python
df.hist(figsize=(12, 10), bins=70, edgecolor='black')
plt.suptitle('Histograms of Numeric Features')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Age'], color='skyblue')
plt.title('Boxplot of Age')
plt.xlabel('Age')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap= 'Oranges', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Titanic Features')
plt.show()
```

```python
subset = df[['Age', 'Fare', 'Pclass', 'SibSp', 'Parch', 'Survived']]
subset = subset.dropna()
sns.pairplot(subset, hue='Survived', palette='husl')
plt.suptitle("Pairplot of Titanic Dataset Features", y=1.02)
plt.show()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
sns.boxplot(x='Survived', y='FamilySize', data=df)
```
